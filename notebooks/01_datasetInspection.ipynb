```python
import numpy as np 
import pandas as pd 
from pathlib import Path
import matplotlib.pyplot as plt


#loading datatset
PROJECT_ROOT = Path("..")
DATA_PATH = PROJECT_ROOT/"data"/"Round10_traces.csv"
df = pd.read_csv(DATA_PATH)

```

```python
#inital inspection
print("Shape: ",df.shape)
print("Columns :",df.columns.tolist())
print("Info :",df.info())
print("Description :",df.describe())
```

```python
#displaying first few rows of datatset
for col in df.columns:
    print(col,df[col].dtype)

pd.set_option("display.max_columns",20)
display(df.head())

#checking missing values
missing = df.isnull().sum()
missing[missing>0]
```

```python
#inspecting columns
df.iloc[:,0].head(10)
print(df.iloc[:, 0].str.len().value_counts()) #length of first column

#trace columns
trace_columns = df.columns[1:] #traces start from 1st column
traces = df[trace_columns].to_numpy() #converting to numpy array
print(traces.shape)

#plotting 1 single trace
plt.figure(figsize=(14, 4))
plt.plot(traces[0])
plt.xlabel("Sample")
plt.ylabel("Power")
plt.title("Trace 0")
plt.show()

#plotting first 10 traces
plt.figure(figsize=(14, 5))
for i in range(10):
    plt.plot(traces[i], alpha=0.5) #alpha is for 50% transparency
plt.xlabel("Sample")
plt.ylabel("Power")
plt.title("First 10 Power Traces")
plt.show()

```

```python
#understanding whether 0th column is ciphertext or plaintext

hex_values = df.iloc[:, 0]

print("Number of unique values:", hex_values.nunique())
print("First 10 values:")

for x in hex_values.head(10):
    print(x)

byte_values = np.array([
    list(bytes.fromhex(x))
    for x in hex_values
], dtype=np.uint8)

print("Shape:", byte_values.shape)
print(byte_values[:5])
```
